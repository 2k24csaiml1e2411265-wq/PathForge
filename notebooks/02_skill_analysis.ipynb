## Top 15 highest-demand skills across all postings

```python
demand = top_demand_skills(jobs_df, top_n=15)
pd.DataFrame(demand, columns=['skill', 'demand_score'])
```

## Skill transferability across roles
Higher score = the skill appears across more DISTINCT job titles (i.e. more transferable if you pivot roles).

```python
transfer = compute_transferability_scores(jobs_df)
pd.DataFrame(sorted(transfer.items(), key=lambda kv: kv[1], reverse=True), columns=['skill', 'transferability_score']).head(15)
```
